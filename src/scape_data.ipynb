```python
import requests
from pypdf import PdfReader
from bs4 import BeautifulSoup
import trafilatura
```

## Hàm lấy url links chứa tài liệu

```python
URL_LINKS="../data/urls.txt"

def get_all_links():
    urls = []
    with open(URL_LINKS, "r", encoding="utf-8") as f:
        for line in f:
            urls.append(line.strip())

    return urls
```

## Collect dữ liệu 

```python
urls = get_all_links()
documents = []

for url in urls:
    downloaded = trafilatura.fetch_url(url)
    # Chuyển đổi sang định dạng Markdown để giữ cấu trúc heading
    content = trafilatura.extract(downloaded, output_format="markdown")
    if content:
        print(f"Get data successfully with: {url}")
        documents.append({"content": content, "metadata": {"source": url}})
    else:
        print(f"Content empty: {url}")
```

## Ghi ra file document.json

```python
import json

with open("../data/document.json", 'w', encoding='utf-8') as f:
    json.dump(documents, f, indent=4)
```
