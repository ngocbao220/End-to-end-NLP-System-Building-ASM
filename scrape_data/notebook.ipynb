```python
import requests
from pypdf import PdfReader
from bs4 import BeautifulSoup
```

```python
def get_doc_type(url):
    prefix = url.split('.')[-1].lower()
    return 'pdf' if prefix == 'pdf' else 'html'

def get_data(url):
    # Wikipedia requires a descriptive User-Agent header
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    doc_type = get_doc_type(url)
    if doc_type == 'html':
        response = requests.get(url, headers=headers)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, "lxml")
            # print(soup)
            header = soup.head
            title = header.title.string if header and header.title else 'No Title'
            print(f"TITLE: {title}")
            tags = soup.find_all('p')
            for tag in tags:
                text = tag.get_text(strip=True)
                if text:
                    print(text)
        else:
            print(f"Failed to fetch {url}, status code: {response.status_code}")
    elif doc_type == 'pdf':
        pass
```

```python
sample_url = 'https://en.wikipedia.org/wiki/List_of_museums_in_Pittsburgh'

get_doc_type(sample_url)

get_data(sample_url)
```
